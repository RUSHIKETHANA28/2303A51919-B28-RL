```python
import gymnasium as gym
import random

# List of environments
environments = ["CliffWalking-v1", "Taxi-v4"] # Changed from "Taxi-v3"

# Number of episodes
num_episodes = 5

# Run the Random Agent for each environment
for env_name in environments:

    print("\n==============================")
    print("Environment:", env_name)
    print("==============================")

    # 1. Create the environment
    env = gym.make(env_name)

    episode_rewards = []

    # Repeat for multiple episodes
    for episode in range(num_episodes):

        # 2. Reset the environment
        state, info = env.reset()

        total_reward = 0
        steps = 0

        terminated = False
        truncated = False

        # Continue until episode terminates
        while not (terminated or truncated):

            # 3. Observe current state
            # state contains the current state

            # 4. Identify available actions
            action = env.action_space.sample()

            # 5 & 6. Select and execute random action
            next_state, reward, terminated, truncated, info = env.step(action)

            # 7. Observe next state and reward
            total_reward += reward
            steps += 1

            # Move to next state
            state = next_state

        # 8 & 9. Episode completed
        average_reward = total_reward / steps

        episode_rewards.append(total_reward)

        print("Episode:", episode + 1)
        print("Steps:", steps)
        print("Total Reward:", total_reward)
        print("Average Reward:", round(average_reward, 2))

    # Average reward across all episodes
    overall_average = sum(episode_rewards) / num_episodes

    print("\nAverage Episode Reward:",
          round(overall_average, 2))

    # Close environment
    env.close()
```
